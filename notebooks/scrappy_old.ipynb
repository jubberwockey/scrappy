```python
def get_portfolio_weights(self, isins, total=1, method='spearman'):
    """
    Get weights for investing in portfolio of isins through deviations from 1/n investment heuristic.
    Highly correlated timeseries will get lower weights.
    
    Returns:
        pd.Series
    """
    corr = self.select_ts(isins).corr(method=method)
    resid_corr = len(corr) - corr.sum()
    weights = resid_corr/resid_corr.sum()
    
    return weights * total

get_portfolio_weights(s, new_favs, 100)
```

```python
def rank_funds(scrappy, criteria=None, timespan='1Y', sparplan_only=True, min_criteria=None):
    """
    Rank funds based on different performance and risk criteria.
    
    Parameters
    ----------
    scrappy: Scrappy
        Scrappy instance with loaded funds_overview data
    
    criteria: dict, optional
        Dictionary of criteria and weights for ranking
        Example: {'performanceTimeSpanPct': 1, 'sharpeRatio': 2, 'volatility': -1, 'TER': -1}
        Positive weights prioritize higher values, negative weights prioritize lower values
        If None, uses a default balanced criteria
    
    timespan: str, default='1Y'
        Time period for evaluation ('1M', '3M', '1Y', '3Y', '5Y', '10Y')
    
    sparplan_only: bool, default=True
        If True, only include funds with sparplan_finanzen_zero = True
    
    min_criteria: dict, optional
        Dictionary of minimum thresholds for criteria
        Example: {'performanceTimeSpanPct': 5, 'sharpeRatio': 0.5, 'TER': 0}
    
    Returns
    -------
    pandas.DataFrame
        Ranked funds with scores and criteria values
    """
    # Default criteria if none provided
    if criteria is None:
        criteria = {
            'performanceTimeSpanPct': 1,    # Higher performance is better
            'sharpeRatio': 2,               # Higher Sharpe ratio is better (weighted 2x)
            'volatility': -0.5,             # Lower volatility is better (weighted 0.5x)
            'maxDrawdown': -0.5,            # Lower drawdown is better (weighted 0.5x)
            'TER': -0.7                     # Lower expense ratio is better
        }
    
    # Default minimum criteria if none provided
    if min_criteria is None:
        min_criteria = {
            'performanceTimeSpanPct': 0,    # Performance should be positive
            'sharpeRatio': 0.3,             # Sharpe ratio should be at least 0.3
            'TER': 0,                       # TER should be available (not missing)
        }
    
    # Filter funds with savings plan if required
    if sparplan_only:
        funds_df = scrappy.funds_overview[scrappy.funds_overview['sparplan_finanzen_zero'] == True].copy()
    else:
        funds_df = scrappy.funds_overview.copy()
    
    # Prepare results DataFrame
    result_df = pd.DataFrame(index=funds_df.index)
    
    # Extract performance and risk metrics for the specified timespan
    for isin in funds_df.index:
        # Check if performance and risk columns exist
        if 'performance' not in funds_df.columns or 'risk' not in funds_df.columns:
            continue
            
        # Check if the columns are lists/arrays and not missing
        if not isinstance(funds_df.loc[isin, 'performance'], list) or not isinstance(funds_df.loc[isin, 'risk'], list):
            continue
        
        # Extract performance data for this timespan
        perf_data = next((item for item in funds_df.loc[isin, 'performance'] 
                          if item['timeSpan'] == timespan), None)
        
        # Extract risk data for this timespan
        risk_data = next((item for item in funds_df.loc[isin, 'risk'] 
                          if item['timeSpan'] == timespan), None)
        
        # Skip if data for this timespan is missing
        if perf_data is None or risk_data is None:
            continue
        
        # Add performance data
        for k, v in perf_data.items():
            if k != 'timeSpan' and k != 'nameTimeSpan':
                result_df.at[isin, k] = v
        
        # Add risk data
        for k, v in risk_data.items():
            if k != 'timeSpan' and k != 'nameTimeSpan':
                result_df.at[isin, k] = v
        
        # Add fund name and TER
        if 'name' in funds_df.columns:
            result_df.at[isin, 'name'] = funds_df.at[isin, 'name']
        if 'TER' in funds_df.columns:
            result_df.at[isin, 'TER'] = funds_df.at[isin, 'TER']
    
    # Drop rows that don't meet minimum criteria
    for criterion, min_value in min_criteria.items():
        if criterion in result_df.columns:
            # Skip NaN values for TER if the criterion is TER
            if criterion == 'TER':
                result_df = result_df[result_df[criterion].notna()]
            else:
                result_df = result_df[result_df[criterion] >= min_value]
    
    # Calculate score based on criteria
    result_df['score'] = 0
    for criterion, weight in criteria.items():
        # Skip if the criterion doesn't exist or all values are NaN
        if criterion not in result_df.columns or result_df[criterion].isna().all():
            continue
            
        # Fill NaN values with the median for scoring purposes
        # This is important for columns like TER which might not be available for all funds
        column_for_calc = result_df[criterion].fillna(result_df[criterion].median())
        
        # Get min and max values for normalization
        min_val = column_for_calc.min()
        max_val = column_for_calc.max()
        
        # Skip normalization if min and max are the same (no variation)
        if max_val == min_val:
            continue
            
        # Normalize values based on whether higher or lower is better
        if weight > 0:  # Higher is better (e.g., performance, Sharpe ratio)
            # Formula: (value - min) / (max - min)
            # This gives 1 for the maximum value and 0 for the minimum
            normalized_score = weight * (column_for_calc - min_val) / (max_val - min_val)
        else:  # Lower is better (e.g., volatility, TER)
            # Formula: (max - value) / (max - min)
            # This gives 1 for the minimum value and 0 for the maximum
            normalized_score = abs(weight) * (max_val - column_for_calc) / (max_val - min_val)
            
        result_df['score'] += normalized_score
    
    # Sort by score (descending)
    result_df.sort_values('score', ascending=False, inplace=True)
    
    # Return funds with relevant columns
    columns_to_include = ['name'] if 'name' in result_df.columns else []
    columns_to_include += [c for c in criteria.keys() if c in result_df.columns] + ['score']
    
    # Make sure we have some columns
    if not columns_to_include:
        return pd.DataFrame()
        
    return result_df[columns_to_include]

# Different ranking strategies
def growth_strategy(s):
    """Rank funds optimized for high growth, tolerating higher volatility"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 2,    # Higher performance heavily weighted
        'sharpeRatio': 1,               # Consider risk-adjusted returns
        'volatility': -0.2              # Accept some volatility
    }, timespan='3Y')

def balanced_strategy(s):
    """Rank funds with balanced risk-return profile"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 1,    # Consider performance
        'sharpeRatio': 2,               # Strongly favor risk-adjusted returns
        'volatility': -0.5,             # Penalize high volatility
        'maxDrawdown': -0.5             # Penalize large drawdowns
    }, timespan='3Y')

def conservative_strategy(s):
    """Rank funds with focus on preserving capital"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 0.5,  # Still want some performance
        'sharpeRatio': 1,               # Consider risk-adjusted returns
        'volatility': -1,               # Strongly penalize volatility
        'maxDrawdown': -1.5             # Heavily penalize drawdowns
    }, timespan='3Y', min_criteria={
        'performanceTimeSpanPct': 0,
        'sharpeRatio': 0.5,
        'maxDrawdown': -30              # Limit maximum drawdown
    })

def short_term_strategy(s):
    """Rank funds for short-term investment horizon"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 1,
        'sharpeRatio': 1
    }, timespan='1Y')

def long_term_strategy(s):
    """Rank funds for long-term investment horizon"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 1,
        'sharpeRatio': 1
    }, timespan='5Y')


# new strategies

def growth_strategy(s):
    """Rank funds optimized for high growth, tolerating higher volatility"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 2.0,  # Performance is the primary focus (2x weight)
        'sharpeRatio': 1.0,             # Consider risk-adjusted returns but not as critical
        'volatility': -0.3,             # Minor penalty for volatility - willing to accept it for growth
        'maxDrawdown': -0.5,            # Some concern about large drawdowns, but will accept moderate ones
        'TER': -0.7                     # Important to keep costs low for long-term compounding
    }, timespan='3Y')
    # Reasoning: Growth investors prioritize high returns, accepting more risk,
    # but still want to avoid excessive fees that eat into returns

def balanced_strategy(s):
    """Rank funds with balanced risk-return profile"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 1.0,  # Performance matters, but balanced with other factors
        'sharpeRatio': 2.0,             # Risk-adjusted return is the primary focus (2x weight)
        'volatility': -0.7,             # More concerned about volatility than growth investors
        'maxDrawdown': -1.0,            # Significant concern about drawdowns
        'TER': -0.8                     # Strong focus on keeping costs low
    }, timespan='3Y')
    # Reasoning: Balanced investors seek optimal risk-adjusted returns with
    # strong emphasis on avoiding deep drawdowns while keeping costs contained

def conservative_strategy(s):
    """Rank funds with focus on preserving capital"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 0.5,  # Performance is less important than safety
        'sharpeRatio': 1.0,             # Risk-adjusted returns are important
        'volatility': -1.5,             # Heavy penalty for volatility (1.5x weight)
        'maxDrawdown': -2.0,            # Very heavy penalty for drawdowns (2x weight) - capital preservation is key
        'TER': -0.7                     # Important to keep costs low, but secondary to safety
    }, timespan='3Y', min_criteria={
        'performanceTimeSpanPct': 0,    # Still require positive performance
        'sharpeRatio': 0.5,             # Minimum acceptable risk-adjusted return
        'maxDrawdown': -25,             # Strict limit on maximum allowed drawdown
        'TER': 1.5                      # Maximum acceptable expense ratio
    })
    # Reasoning: Conservative investors prioritize capital preservation above all,
    # with maximum drawdown as their primary concern

def short_term_strategy(s):
    """Rank funds for short-term investment horizon (1 year)"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 1.5,    # Recent performance is highly important
        'sharpeRatio': 1.2,               # Risk-adjusted returns are important
        'volatility': -0.8,               # Higher penalty for volatility in short-term
        'maxDrawdown': -1.0,              # Short-term investors are sensitive to drawdowns
        'TER': -0.5                       # Fees matter less over shorter timeframes
    }, timespan='1Y', min_criteria={
        'performanceTimeSpanPct': 0,      # Should have positive performance
        'sharpeRatio': 0.2,               # Minimum risk-adjusted return
        'maxDrawdown': -20                # Limit maximum drawdown for short-term safety
    })
    # Reasoning: Short-term investors prioritize recent performance and 
    # protection against sudden drawdowns, with moderate concern for costs
    # since fee impact is less significant over shorter periods

def long_term_strategy(s):
    """Rank funds for long-term investment horizon (5+ years)"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 1.2,    # Long-term performance is important
        'sharpeRatio': 1.5,               # Strong emphasis on risk-adjusted returns
        'volatility': -0.4,               # Less concerned about volatility given long horizon
        'maxDrawdown': -0.7,              # Moderate concern for drawdowns
        'TER': -2.0                       # Very heavy penalty for high fees (2x weight)
    }, timespan='5Y', min_criteria={
        'performanceTimeSpanPct': 2,      # Require meaningful long-term performance
        'TER': 1.8                        # Strict limit on maximum expense ratio
    })
    # Reasoning: Long-term investors benefit most from fee compounding effects,
    # so TER receives the highest weight. Performance consistency over 5 years
    # is valued over short-term metrics. Lower sensitivity to volatility
    # due to longer time horizon for recovery

def income_strategy(s):
    """Rank funds focused on generating stable income"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 0.8,  # Performance matters, but stability is more important
        'sharpeRatio': 1.5,             # Strong emphasis on risk-adjusted returns
        'volatility': -1.2,             # Heavy penalty for volatility
        'maxDrawdown': -1.0,            # Significant concern about drawdowns
        'TER': -0.8                     # Strong focus on keeping costs low to preserve income
    }, timespan='3Y', min_criteria={
        'performanceTimeSpanPct': 1,    # Require at least 1% performance
        'TER': 1.2                      # Maximum acceptable expense ratio
    })
    # Reasoning: Income investors want stable returns with minimal volatility,
    # focusing on consistent performance rather than maximum growth

def esg_focus_strategy(s):
    """Rank funds with strong ESG metrics (assuming ESG rating data exists)"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 0.8,  # Performance still matters
        'sharpeRatio': 1.0,             # Risk-adjusted returns are important
        'volatility': -0.5,             # Some concern about volatility
        'maxDrawdown': -0.5,            # Some concern about drawdowns
        'TER': -0.5,                    # Willing to accept slightly higher fees for ESG focus
    }, timespan='3Y', sparplan_only=True)
    # Reasoning: ESG investors balance financial performance with sustainability goals,
    # often accepting slightly higher fees or lower returns for alignment with values

def tactical_allocation_strategy(s):
    """Rank funds ideal for tactical allocation based on current market conditions"""
    return rank_funds(s, criteria={
        'performanceTimeSpanPct': 1.5,  # Recent performance is important indicator
        'sharpeRatio': 0.7,             # Consider risk-adjusted returns
        'volatility': -0.3,             # Low concern about volatility
        'maxDrawdown': -0.3,            # Low concern about historical drawdowns
        'TER': -0.5                     # Some concern about costs
    }, timespan='1Y', sparplan_only=False)
    # Reasoning: Tactical investors focus on recent performance and momentum,
    # with less concern about historical risk metrics


balanced_funds = balanced_strategy(s)
growth_funds = growth_strategy(s)
conservative_funds = conservative_strategy(s)
short_term_funds = short_term_strategy(s)
long_term_funds = long_term_strategy(s)
income_funds = income_strategy(s)



print("balanced")
display(balanced_funds.head(10))
print("growth")
display(growth_funds.head(10))
print("conservative")
display(conservative_funds.head(10))
print("short-term")
display(short_term_funds.head(10))
print("long-term")
display(long_term_funds.head(10))
print("income")
display(income_funds.head(10))
```

```python
import plotly.express as px
# import plotly.graph_objects as go

def get_ts_long(ts_wide, id_var='x'):
    fonds_names = list(ts_wide.columns)
    fonds_names.remove(id_var)
    return pd.melt(ts_wide, id_vars=[id_var], value_vars=fonds_names, var_name='name', value_name='y')

def plot_ts(ts):
    fig = px.line(get_ts_long(ts), x='x', y='y', color='name')
    # fig.update_layout(legend={'yanchor': 'top', 'xanchor': 'left', 'y': 1, 'x': 0})
    fig.update_layout(showlegend=False)
    fig.show()
```
